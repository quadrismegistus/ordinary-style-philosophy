```python
import sys; sys.path.append('..')
from osp import *
```

```python
df = get_all_feats(normalize=True)
df
```

```python
from sklearn.decomposition import PCA

n=10
pca = PCA(n_components=n)
df_pca = pca.fit_transform(df)
df_pca = pd.DataFrame(df_pca, columns=[f'PC{i+1}' for i in range(n)], index=df.index)
df_pca
```

```python
# Show factor loadings (feature contributions to each principal component)
n_components = pca.n_components_
pc_cols = [f'PC{i+1}' for i in range(n_components)]

loadings = pd.DataFrame(
    pca.components_.T,
    columns=pc_cols,
    index=df.columns
)

# Add absolute value columns for each PC
for pc_col in pc_cols:
    loadings[f'abs_{pc_col}'] = loadings[pc_col].abs()

# Print top features for each PC
for pc_col in pc_cols:
    ldf=loadings.sort_values(f'abs_{pc_col}', ascending=False)[[pc_col, f'abs_{pc_col}']].head(20)
    if 'sent_IC' not in ldf.index: continue
    print(f"Top features contributing to {pc_col}:")
    print(', '.join(ldf.index.tolist()))
    print()

# loadings

```
