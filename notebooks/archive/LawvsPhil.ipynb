```python
import sys
sys.path.append('..')
from osp import *
```

```python
df = STASH_SLICE_FEATS.df
```

```python
df_meta = get_corpus_metadata()
id2journal = dict(zip(df_meta.index, df_meta.journal))

df['journal'] = [id2journal.get(i.split('__')[0],'') for i in df.index]
```

```python
pd.set_option('display.max_rows', 500)

df['law_or_phil'] = ['Law' if ' Law ' in j else ('Philosophy' if i.startswith('phil/') else 'Neither') for i,j in zip(df.index,df.journal)]

df['pos_MD'] = (df['pos_MD'] - df['pos_MD'].mean()) / df['pos_MD'].std()
```

```python
df['deprel_mark'] = (df['deprel_mark'] - df['deprel_mark'].mean()) / df['deprel_mark'].std()
```
