```python
df_feats = pd.read_pickle('../data/raw/df_feats3.pkl.gz')
df_feats['weight_abs'] = df_feats['weight'].abs()
df_feats = df_feats.groupby('feature').mean(numeric_only=True).reset_index().drop(columns=['run']).sort_values('weight_abs',ascending=False)
```

```python
df_feats1 = pd.read_pickle('../data/raw/df_feats3.pkl.gz')
df_feats1['weight_abs'] = df_feats1['weight'].abs()
df_feats1 = df_feats1[df_feats1.comparison.str.contains('Philosophy') & df_feats1.comparison.str.contains('Other')].groupby('feature').mean(numeric_only=True).reset_index().drop(columns=['run']).sort_values('weight_abs',ascending=False)
df_feats2 = df_feats1.copy()
df_feats2['feat_type'] = df_feats2['feature'].apply(lambda x: x.split('_')[0])
df_feats2['num'] = 1
```

```python
df_feats2.groupby('feat_type').agg(
    {
        'num': 'sum',
        'weight_abs': 'max',
    }
).sort_values('num',ascending=False)
```

```python
figdf = df_feats.drop(columns=['weight_abs','weight']).melt(id_vars=['feature'])
figdf['variable'] = figdf['variable'].apply(lambda x: x.split("_")[-1])
variables = figdf['variable'].unique().tolist()
discs = ['Philosophy','Literature','Other']
variables.sort(key=lambda x: (discs.index(x.split()[-1]), x.split()[0]))

feats = figdf[figdf['variable'].str.contains('Philosophy')].groupby('feature').mean(numeric_only=True).sort_values('value', ascending=True).index
feats
```

```python
x=defaultdict(list)
for k,v in FEAT2DESC.items():
    x[v].append(k)

for k,v in x.items():
    if len(v) > 1:
        print(k,v)
```

```python
figdf['feature'] = pd.Categorical(figdf['feature'], categories=feats)#df_feats.groupby('feature').mean(numeric_only=True).sort_values('weight_abs', ascending=True).index)
figdf['feature2'] = figdf['feature'].apply(lambda x: f'{FEAT2DESC[x]} ({x.split("_")[0]})')
# figdf['feature2'] = figdf['feature'].apply(lambda x: f'{FEAT2DESC[x]}')
# figdf['feature2'] = pd.Categorical(figdf['feature2'], categories=[FEAT2DESC[x] for x in feats])
figdf['variable'] = pd.Categorical(figdf['variable'], categories=variables)
figdf['value_str'] = figdf['value'].apply(lambda x: f'+{x:.1f}' if x > 0 else f'{x:.1f}')
figdf['value'] = figdf['value'].apply(lambda x: 1 if x>1 else x)
figdf['value'] = figdf['value'].apply(lambda x: -1 if x<-1 else x)

fig = (
    p9.ggplot(
        figdf,
        p9.aes(
            y='feature2',
            x='variable',
            fill='value',
            label='value_str'
        )
    )
    + p9.geom_tile()
    + p9.geom_text(
        size=6
    )
    + p9.theme_minimal()
    + p9.theme(
        axis_text_x=p9.element_text(angle=45, hjust=1),
        figure_size=(10, 14),
        plot_background=p9.element_rect(fill='white'),
        panel_background=p9.element_rect(fill='white'),
        plot_title = p9.element_text(size=10)
    )
    + p9.scale_fill_gradient2(
        # low='darkred',
        # high='lightblue',
        limits=(-1,1)
    )
    + p9.labs(
        x='',
        y='',
        fill='Frequency\n(z-score)',
        title='Historical frequency of syntactic features in journals of philosophy, literary criticism, and other disciplines'
    )
    + p9.guides(fill=False)
)
fig.save('../figures/feat_table.png')
fig
```
